# **Python Modules, Packages and Libraries**

**A module** is a file containing Python code that defines functions, variables, and classes. They help to group related functions together, and their contents can be imported and used in other python scripts.

It is a single Python script (a .py file) that contains Python code.

```python
def greet(name):                          # This is a python function with one argument 'name'
    return f"Welcome to {name}!"          # f-string (formatted string)

greet("QMSS")
```

```python
import testing_module

message = testing_module.greet("The univerity of Michigan")
message
```

```python
import testing_module as tm

message = tm.greet("LSA")
message
```

**A library** is a collection of related modules and functions that serve a specific domain or purpose. It typically consists of multiple modules bundled together for a common use case.

Thus, it is imported and treated like a module.

# **Data Structures in Python**

**LIST:** A built-in data structure that allows you to store and organize a collection of items.

They are ordered, indexed, mutable, and heterogeneous.

```python
new_list = [1, 2, 3, 4, 5, 6, 7, 8, 9]
first_element = new_list[0]       # Access the first element (1)
sublist = new_list[2:4]           # Create a sublist [3, 4]

sublist
```

```python
length = len(new_list)
length
```

```python
new_list
```

```python
# Appending a list
new_list.append(10)
new_list
```

```python
# adding a list to another
new_list.extend([11, 12])
new_list
```

```python
# Inserting an element into a list
new_list.insert(3, 13)    # It inserts 13 in the third index
new_list
```

```python
# Removing an element in a list
new_list.remove(2)  # Removed the first 2 in the list
new_list
```

```python
# Removing a list element by index
new_list.pop(0)
new_list
```

**DICTIONARY**: A Python dictionary is an unordered collection of key-value pairs.

It is a versatile data structure that allows us to store, retrieve, and manipulate data efficiently.

```python
# Creating a dictionary. Student is a dictionary with 6 key-value pairs.
student = {
    'name': 'Kelvin',
    'major': 'Pol Sci',
    'level': 'Senior',
    'age': 22,
    'grade': 'A',
    'is_student': True
}
```

```python
# Accessing the value of a dictionary

# Accessing values
print("Name:", student['name'])
print("Age:", student['age'])
```

```python
# Modifying the values

# Modifying values
student['major'] = 'Economics'
student['grade'] = 'B'
student['age'] = 24

student
```

```python
# Adding and removing key-value pairs

student['state'] = 'Michigan'   # Add a new key-value pair
student.pop('is_student')       # Removing a key-value pair
student
```

```python
# Checking if a key exist in a dictionary
if 'major' in student:
    print("Major exists in the dictionary.")
```

```python
# The value can also be a list

Income_data = {'Name': ['Kelvin', 'Korede', 'Alex'],
        'Income': [100000, 120000, 55000]}
Income_data
```

**Dataframe:** A DataFrame is a two-dimensional, tabular data structure commonly used in Python for data manipulation and analysis.

It is particularly associated with the **pandas** library.

```python
# DATAFRAME
import pandas as pd

# Creating a dataframe from a dictionary

Income_data = {'Name': ['Kelvin', 'Korede', 'Alex'],
        'Income': [100000, 120000, 55000]}

Income_df = pd.DataFrame(Income_data)
Income_df
```

```python
# Creating dataframe from list

names = ['Charles', 'Yoyo', 'Nick']
ages = [25, 30, 22]
age_df = pd.DataFrame({'Name': names, 'Age': ages})
age_df
```
