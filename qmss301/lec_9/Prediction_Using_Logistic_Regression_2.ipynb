Risk Level: Predicted Risk Intensity Level during pregnancy considering the previous attribute.

# **Using Statmodel**

*1.3 Training & Testing Split Strategy*

```python
# This is setting up what are our independent variables and dependent variable of the ML model
x = maternal  # Explanatory variables (IVs)
y = maternal['RiskLevel1']  # Response variable (DV)
```

**Nemo iudex in causa sua**

Latin for “no one should be a judge in their own cause”. It is one of the cardinal rules of natural justice that no one should act as a judge a case in which they have a personal (vested) interest.

```python
# FOR NOW, LET'S BORROW THE 'train_terst_split' FUNCTION FROM SKLEARN

from sklearn.model_selection import train_test_split

# Spliting the data into training data and testing data
# The training dataset is used to build the logistic model, while the testing dataset is used to evaluate the model.

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    train_size = 0.8, # This is the proportion of the dataset to include in the train split.
    random_state = 16, # This is set to control the shuffling applied to the data before spliting it. It helps to make the work reproducible
)
```

```python
# Looking at the sizes of the training and testing data samples
print("Size of Training Samples: ", len(y_train))
print("Size of Testing Samples: ", len(y_test))
```

```python
# Within both testing and training data, let us look at how many are there for high risk or low risk outcomes.

# Testing data
print("Number of observations for the 'Low Risk' outcome in test data: ", (y_test == 0).sum())
print("Number of observations for the 'High Risk' outcome in test data: ", (y_test == 1).sum())

# Training Data.
print("Number of observations for the 'Low Risk' outcome in train data: ", (y_train == 0).sum())
print("Number of observations for the 'High Risk' outcome in train data: ", (y_train == 1).sum())
```

```python
(y_test == 0).sum()
```

**2. RUNNING THE REGRESSION**

```python
from statsmodels.formula.api import logit

var_struct = "RiskLevel1 ~ BS + HeartRate"

# Here, we look at a multivariate logistic regression
mat_model =logit(var_struct, data = x_train).fit()
print(mat_model.summary())
```

**3. MAKING PREDICTIONS WITH THE MODEL**

```python
expl_data = x_test[["BS", "HeartRate"]] # Subsetting the column of the explanatory variables
x_test['pred'] = mat_model.predict(expl_data)
x_test.head()
```

```python
x_test.shape
```

The approximate manual way of doing this is:

```python
x_test["prob"] = 1/(1+np.exp(-(-12.0156 + 1.1747*x_test["BS"] + 0.0266*x_test["HeartRate"])))
x_test.head()
```

**4. EVALUATING THE MODEL**

```python
# prob > 0.5 means 1 (High Risk)
# prob < 0.5 means 0 (Low Risk)
x_test["prediction"]=np.where(x_test["pred"] > 0.5, 1, 0)
x_test[:5]
```

Shape.png

```python
# Finding accuracy by comparing the predicted to the actual.
x_test["pred2true"]= np.where(x_test["prediction"] == x_test["RiskLevel1"],'Right','Wrong')
x_test.head()
```

```python
# ALSO, LET US BORROW THE CONFUSION MATRIX FUNCTION FROM SKLEARN
from sklearn.metrics import confusion_matrix

actual = x_test["RiskLevel1"] # The actual response
predicted = x_test["prediction"] # The predicted response

# Create a confusion matrix
cm = confusion_matrix(actual, predicted)

cm
```

```python
print("True Prediction: ", len(x_test.query("pred2true == 'Right'")))
print("False Prediction: ", len(x_test.query("pred2true == 'Wrong'")))
print("Prediction Accuracy: ", len(x_test.query("pred2true == 'Right'"))/len(y_test))
```

```python
# Create a heatmap using seaborn and matplotlib libraries
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(cm, annot=True, cmap='Blues', cbar=True)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

# **Using sklearn**

```python
# Select features (X) and the target variable (y)
X = maternal[['BS', 'HeartRate']]
y = maternal['RiskLevel1']
```

```python
# Kindly take note of how X and Y are subsetted between this and when we use statmodel
# For statmodel

# x = maternal  # Explanatory variables (IVs)
# y = maternal['RiskLevel1']  # Response variable (DV)
```

```python
from sklearn.model_selection import train_test_split

# Spliting the data into training data and testing data
# The training dataset is used to build the logistic model, while the testing dataset is used to evaluate the model.

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    train_size = 0.8, # This is the proportion of the dataset to include in the train split. If none, it will be automatically set to 0.25
    random_state = 16, # This is set to control the shuffling applied to the data before spliting it. It helps to make the work reproducible
)
```

```python
from sklearn.linear_model import LogisticRegression

# Create a Logistic Regression Classifier
clf = LogisticRegression()

# Train the model
clf.fit(X_train, y_train)

# Access and print the coefficients and intercept
print("Coefficients:", clf.coef_)
print("Intercept:", clf.intercept_)

```

```python
# Make predictions on new data (X_test)
y_pred = clf.predict(X_test)
```

```python
# Access and print the accuracy on the training data
train_accuracy = clf.score(X_test, y_test)
print("Training accuracy:", train_accuracy)
```

Shape.png

```python
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix

# Create a confusion matrix
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
# Create a heatmap
sns.heatmap(cm, annot=True, fmt='d',cmap='Blues', cbar=False)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

**Model Prediction Accuracy**

We will talkk about 4 different model prediction accuracy for classification problems:

1. Accuracy score

2. precision score(Positive Predictive Value):

3. recall score

4. F1 score


```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, roc_curve

# Make predictions on the test data
y_pred = clf.predict(X_test)
```

1. **Accuracy score:** Accuracy measures the ratio of correctly predicted instances to the total number of instances (sample size). It provides an overall assessment of how well the model performs.

  **Formula:** (True Positives + True Negatives) / (True Positives + True Negatives + False Positives + False Negatives).



```python
# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

2. **Precision (Positive Predictive Value):** Precision measures the ability of the model to correctly identify positive instances (e.g., classifying an email as spam when it is actually spam). It's a measure of the model's precision in making positive predictions.

  **Formula:** True Positives / (True Positives + False Positives).



```python
# Calculate precision
precision = precision_score(y_test, y_pred)
print("Precision:", precision)
```

3. **Recall (Sensitivity, True Positive Rate):** It measures the ability of the model to correctly capture all positive instances. It is particularly important when missing a positive instance has significant consequences (e.g., in medical diagnoses).

  **Formula:** True Positives / (True Positives + False Negatives)


```python
# Calculate recall
recall = recall_score(y_test, y_pred)
print("Recall:", recall)
```

4. **F1-Score:** The F1-Score is the harmonic mean (reciprocal of the arithmetic mean of the reciprocals of a set of values) of precision and recall. It provides a balanced measure of precision and recall, which is useful when you want to find a balance between the two. It is especially valuable in imbalanced datasets.

  **Formula:** 2 * (Precision * Recall) / (Precision + Recall)


```python
# Calculate F1-score
f1 = f1_score(y_test, y_pred)
print("F1-Score:", f1)
```

```python
from sklearn.metrics import classification_report

# Print a classification report
report = classification_report(y_test, y_pred)
print("Classification Report:\n", report)
```

Macro Average (Macro Avg): The macro average is the unweighted average of precision, recall, and F1-score across all classes. It provides an overall summary of the model's performance without considering class imbalances.

Weighted Average (Weighted Avg): The weighted average is similar to the macro average but takes into account class imbalances by weighting each class's performance by its support. It provides an overall summary that considers the distribution of classes in the dataset
