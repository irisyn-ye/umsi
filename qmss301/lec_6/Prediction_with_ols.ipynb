```python
# Importing the necessary Python Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.formula.api import ols
from itertools import product
```

**1. RUN THE REGRESSION MODEL**

```python
# reading the csv file
car_data = pd.read_csv("auto.csv")

# listing the last five rows of tabular data
car_data.tail()
```

```python
# Looking at the effects of weight and cylinders in miles per galon(mpg) and printing out the result
car_multi_reg = ols('mpg ~ weight + cylinders', data = car_data).fit()
print(car_multi_reg.summary())
```

**2. USING THE MODEL TO MAKE PREDICTIONS**

*2.1 Making the predictions using data from the original dataset*

```python
# Using .fittedvalues
print(car_multi_reg.fittedvalues)                         # Print out the fitted values
car_data['prediction'] = car_multi_reg.fittedvalues       # Create a column for the fitted values
car_data.head(5)
```

This is also equivalent to

```python
exp_orig = car_data[['weight','cylinders']]
main_pred = car_multi_reg.predict(exp_orig)
print(main_pred)
```

```python
# Comparing the actual values and predicted values
compare = car_data[['mpg','prediction']]
compare
```

**3. THE RESIDUALS**

This is the difference between the actual value and the value predicted using the regression model.

```python
print(car_multi_reg.resid) # Calling the .resid on the actual regression model
```

We can also calculate this manually

```python
# This is the real actual response variable 'mpg' minus the predictions from the fitted value
print(car_data['mpg'] - car_multi_reg.fittedvalues)
```

**4. LOOKING AT THE MODEL'S FITNESS**

This shows how well our model, with its estimated parameters, fits the data.

We are going to consider two measures of it here:

*4.1 R-Squared (Coefficent of determination)*: Measures the proportion of variance in the response variable that can be explained by the explanatory variable(s). It ranges from 0 (poor fit) to 1 (perfect fit). Therefore the closer it is to 1, the better our model's fitness.

*4.2 Residual standard error (RSE)*: This is equal to square root of mean squared error(MSE). It is the difference between the observed response and the prediction. The smaller it is the better.

```python
# Calculating the R-Squared
print(car_multi_reg.rsquared)
```

The result shows that about 69% of the variation in our response variable (mpg) can be explained by our explanatory variables (weight and cylinders)

```python
print(car_multi_reg.rsquared_adj)
```

```python
# Calculating the RSE
mse = car_multi_reg.mse_resid
rse = np.sqrt(mse)
print(rse)
```

This means that the difference between the actual mpg and predicted mpg is about 4.32.
